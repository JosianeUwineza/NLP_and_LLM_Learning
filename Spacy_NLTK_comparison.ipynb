### Tokenization
Tokenization is a process of splitting text into meangfull segments.

```python
import spacy

nlp = spacy.blank("en")
text = 'Dr. Strange loves pav bhaji of mumbai as it costs only 2$ per plate'
doc = nlp(text)
for token in doc:
    print(token)
```

```python
student_tex_path = '/home/ircad/Documents/Learning_text_classification/NLP_datasets/student.txt'

with open(student_tex_path) as f:
    text = f.readlines()

text = " ".join(text)
student_doc = nlp(text)
emails = []
for token in student_doc:
    if token.like_email:
        emails.append(token)

print(emails)
```

### Exercises

```python
par_text = '''
Look for data to help you address question. Governements are good sources because data from public research is often freely available. Good places to start include http://www.data.gov/, and http://www.science.gov/, and in the United Kingdom, http://data.gov.uk/. Two of my favorite data sets are the General Social Survey at http://www3.norc.org/gss+website/, and the European Social Survey at http://www.europeansocailsurvey.org/.
'''
# Grab all url

par_doc = nlp(par_text)
data_urls = []
for token in par_doc:
    if token.like_url:
        data_urls.append(token)

print(data_urls)
```

```python
transactions = "Tony gave two $ to Peter, Bruce gave 500 € to Steve"
t_doc= nlp(transactions)
for token in t_doc:
    if token.like_num or token.is_currency:
        print(token)
```
