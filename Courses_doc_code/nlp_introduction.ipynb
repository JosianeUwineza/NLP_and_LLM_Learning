### NLP Pipeline: Practical Natural Language Proccessing Book
* Data acquisation 
* Text extraction and Cleanup
* Text proprocessing: Split sentence or sentence tokenization, Word Tokenization, Stemming and Lemmitazion: Mapping word to their roots (eats to eat)
* Feature engineering: Converting word to vectors
* Build Machine learning Models(GridSearchCV techniques: help to find best ML models on certain problem)
* Evaluate Model(precison, recall, f1 score)
* Export and Deploy Models
* Moniter and update the system


### Library used in NLP
* NLTK, Spacy, Gensim, Pytorch, Hugging Face, Scikit-learn. NLTK and SPacy are most popular library that are used in NLP and they tend to do similar things.

| NLTK | SPacy |
|------|-------|
|String processing library| Object oriented processing library|
|Provides access to many algorithms. If you care about specific algo and customozations go with NLTK |Provides most efficient NLP algorithm for a given task. Hence if you care about the end results, go with spacy|
| NLTK is also user friendly but less user freindly compared to Spacy    | Spacy is User friendly|
| Not better results than Spacy   | Better results than NLTK since it new library |
| Better for researchers          | Better for developpers |


```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Dr.Strange loves pav bhaji of mumbai. Hulk loves chaat of delhi")
for sentence in doc.sents:
    print(sentence)
```

```python
for sentence in doc.sents:
    for word in sentence:
        print(word)
```
