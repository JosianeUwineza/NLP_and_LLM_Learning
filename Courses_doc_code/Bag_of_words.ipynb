## Bag of words
* It deals with how many times a certain words appears in text or articles
* For example:

|       |  Musk | Tesla | iphone | ipad | model-3 | cook | itunes  | gigafactory | ....|
|-------|-------|--------|-------|------|---------|------|---------|-------------|-----|
|article1|0    | 14      |0   | 0  |  9 |0     |  0    |  2   |...|
|article2|0    | 0      |13  | 6  |  0 |0     |  3    |  0  |...|
|article3|5    | 7      |0   | 0  |  0 |1    |  0    | 4   |...|
* Having text or doc Vocaburary (unique words in text), by counting the number that a certain vocuburary appears in the text of your dataset this process is called bag of words also it can be called count vectorizer.

## Disadvantages 
* It may consumes too much memory  and computer ressources
* Do not capture meaning of words correctly the meaning of word
