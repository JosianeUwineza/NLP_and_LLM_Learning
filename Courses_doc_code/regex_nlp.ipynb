### Regular Expressions
#### Common token 
| Description| Regex Symbol |
|------------|--------------|
|Single character: a, b , or c | **[abc]** |
|A character except: a, b , or c | **[^abc]** |
|A character in range: a-z |  [a-z] |
|A character in in range: a-z | [^a-z] |
| A character in range: a-z or A-z | [a-zA-Z] |
| Any single character | . |
| Alternate: Match either a or b | **a vertical line b** |
| Any whitespace charater | \s |
| Any non-whitespace charater | \S |
| Any digit | \d |
| Any non-digit | \D|
| Any word character | \w|
| Any non-word character | \W|
| Non-capturing group | (?:....)|
| Capturing group | (....)|
| Zero or one of a | a? |
| Zero or more of a | a* |
|One or more of a | a+ |
| Exactly 3 of a | a{3}|
|3 or more a | a{3, } |
|Between 3 and 6 of a | a{3, 6}|
| Start of srting | ^ |
| End of string  | $ |
| A word boundary | \b |
| Non-word boundary | \B |

#### General Tokens
| Description| Regex Symbol |
|------------|--------------|
|New line | \n |
| Carrige return | \r |
| Tab | \t|
| Null character | \0|

```python
import re

text = " Elon Musk's phone number is 9991116666, call him if you have any questions on dodgecoin. Tesla's revenue is 40 billion. "
pattern = r"\d{10}"
match = re.findall(pattern, text)
print(match )
```

```python
chat1  = 'codebasics: you ask lot of questions 1235678912, abc@xyz.com, 9998881234'
chat2 =  'codebasics: here it is : (123)-567-8912, abX_82@xyz.com'
chat3 = 'codebasics: yes, phone: 1235678912 email: abc@xyz.com'
```

```python
pattern1 = r"\(\d{3}\)-\d{3}-\d{4}| \d{10}"
match = re.findall(pattern1, chat3)
print(match)
```

```python
pattern2 = r"[a-zA-z0-9_]*@[a-zA-z0-9]*\.[a-z]*"
match = re.findall(pattern2, chat2)
print(match)
```

```python
chat4 =" codebasis: Hello, I am having an issue with my order # 412889912"
chat5 =" codebasis: Hello, I have a problem with my order number is 412889912"
chat6 =" codebasis: My order 41282 is having an issue, I was charged 300 $ when online it says 280 $"
```

```python
pattern3 = r"order[^\d]*(\d*)"
match3 = re.findall(pattern3, chat6)
print(match3)
```

```python
def get_pattern_match(pattern, text):
    matches = re.findall(pattern, text)
    if matches:
        return matches[0]
```

```python
get_pattern_match("order[^\d]*(\d*)", chat4)
```
